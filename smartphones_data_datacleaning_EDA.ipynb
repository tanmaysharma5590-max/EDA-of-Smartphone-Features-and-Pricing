## 🧹 Tidiness Issues

1. **SIM** – Can be split into 3 columns:

   * `has_5g`, `has_NFC`, `has_IR_Blaster`
2. **RAM** – Can be split into 2 columns:

   * `RAM`, `ROM`
3. **Processor** – Can be split into:

   * `processor_name`, `cores`, `cpu_speed`
4. **Battery** – Can be split into:

   * `battery_capacity`, `fast_charging_available`
5. **Display** – Can be split into:

   * `size`, `resolution_width`, `resolution_height`, `frequency`
6. **Camera** – Can be split into:

   * `front_camera`, `rear_camera`
7. **Card** – Can be split into:

   * `supported`, `extended_upto`
